크기에 대한 주파수응답을 그래프로 그리시오.

```python
# f_Hz =   [   ??,    ??,    ??,    ??,    ??,    ??,    ??]
# V_V_PP = [   ??,    ??,    ??,    ??,    ??,    ??,    ??]
```

```python
f_Hz = np.array(f_Hz)
V_V_PP = np.array(V_V_PP)

# ---- Circuit parameters ----
R = 1.5 * 1000        # ohm
C = 0.1 / 1000000     # F
fc = 1 / (2 * np.pi * R * C)   # ~1061 Hz
 
# Input amplitude (Vpp)
V_in_V_PP = 4.0
 
# ---- Theoretical curve ----
f_th = np.logspace(np.log10(f_Hz.min() / 2), np.log10(f_Hz.max() * 2), 500)
H_th = 1 / np.sqrt(1 + (f_th / fc) ** 2)
 
gain_meas_dB = 20 * np.log10(V_V_PP / V_in_V_PP)
gain_th_dB = 20 * np.log10(H_th)
 
# Theoretical curve for the linear plot (evenly spaced on a linear axis)
f_th_lin = np.linspace(0, f_Hz.max() * 1.05, 1000)
H_th_lin = 1 / np.sqrt(1 + (f_th_lin / fc) ** 2)
 
# ---- Plot ----
# ax0: linear axes (x is not shared, because a shared x axis must use one scale)
# ax1, ax2: log frequency axis, shared
fig = plt.figure(figsize=(8, 12))
ax0 = fig.add_subplot(3, 1, 1)
ax1 = fig.add_subplot(3, 1, 2)
ax2 = fig.add_subplot(3, 1, 3, sharex=ax1)
 
fig.suptitle(f'1st-order RC LPF (R = 1.5 k$\\Omega$, C = 0.1 $\\mu$F, '
             f'$f_c$ = {fc:.0f} Hz, $V_i$ = {V_in_V_PP:g} V$_{{pp}}$)')
 
# (0) Output amplitude, linear-linear
ax0.plot(f_th_lin, V_in_V_PP * H_th_lin, '-', label='Theory')
ax0.plot(f_Hz, V_V_PP, 'o', label='Measured')
ax0.axvline(fc, color='gray', ls='--', lw=1)
ax0.axhline(V_in_V_PP / np.sqrt(2), color='gray', ls=':', lw=1)
ax0.set_xlim(0, f_Hz.max() * 1.05)
ax0.set_ylim(0, None)
ax0.set_xlabel('Frequency [Hz]')
ax0.set_ylabel('$V_o$ [V$_{pp}$]')
ax0.set_title('Linear scale')
ax0.grid(True, ls=':', alpha=0.6)
ax0.legend()
 
# (1) Output amplitude, log-log
ax1.loglog(f_th, V_in_V_PP * H_th, '-', label='Theory')
ax1.loglog(f_Hz, V_V_PP, 'o', label='Measured')
ax1.axvline(fc, color='gray', ls='--', lw=1)
ax1.axhline(V_in_V_PP / np.sqrt(2), color='gray', ls=':', lw=1)
ax1.set_ylabel('$V_o$ [V$_{pp}$]')
ax1.set_title('Log-log scale')
ax1.grid(True, which='both', ls=':', alpha=0.6)
ax1.legend()
ax1.tick_params(labelbottom=False)
 
# (2) Gain in dB, log frequency (Bode magnitude)
# Straight-line (asymptotic) Bode approximation:
#   f <= fc: 0 dB
#   f >  fc: -20 dB/dec line through 0 dB at fc  -> -20*log10(f/fc)
gain_asym_dB = np.minimum(-20 * np.log10(f_th / fc), 0)
 
ax2.semilogx(f_th, gain_th_dB, '-', label='Theory')
ax2.semilogx(f_th, gain_asym_dB, '-', color='red', lw=1.5,
             label='Straight-line approx.')
ax2.semilogx(f_Hz, gain_meas_dB, 'o', label='Measured')
ax2.axvline(fc, color='gray', ls='--', lw=1, label=f'$f_c$ = {fc:.0f} Hz')
ax2.axhline(-3, color='gray', ls=':', lw=1, label='-3 dB')
ax2.set_xlabel('Frequency [Hz]')
ax2.set_ylabel('Gain [dB]')
ax2.set_title('Bode magnitude (semilog-x)')
ax2.grid(True, which='both', ls=':', alpha=0.6)
ax2.legend()
 
plt.tight_layout()
# plt.savefig('lpf_freq_response.png', dpi=150)
plt.show()
```

```python
f_Hz = np.array(f_Hz)
V_V_PP = np.array(V_V_PP)

# ---- Circuit parameters ----
R = 1 * 1000          # ohm
C = 0.01 / 1000000    # F
fc = 1 / (2 * np.pi * R * C)   # ~15.9 kHz
 
# Input amplitude (Vpp)
V_in_V_PP = 4.0
 
# ---- Theoretical curve ----
# 1st-order RC HPF: |H| = (f/fc) / sqrt(1 + (f/fc)^2)
def H_hpf(f):
    x = f / fc
    return x / np.sqrt(1 + x ** 2)
 
f_th = np.logspace(np.log10(f_Hz.min() / 2), np.log10(f_Hz.max() * 2), 500)
H_th = H_hpf(f_th)
 
gain_meas_dB = 20 * np.log10(V_V_PP / V_in_V_PP)
gain_th_dB = 20 * np.log10(H_th)
 
# Theoretical curve for the linear plot (evenly spaced on a linear axis)
f_th_lin = np.linspace(0, f_Hz.max() * 1.05, 1000)
H_th_lin = H_hpf(f_th_lin)
 
# ---- Plot ----
# ax0: linear axes (x is not shared, because a shared x axis must use one scale)
# ax1, ax2: log frequency axis, shared
fig = plt.figure(figsize=(8, 12))
ax0 = fig.add_subplot(3, 1, 1)
ax1 = fig.add_subplot(3, 1, 2)
ax2 = fig.add_subplot(3, 1, 3, sharex=ax1)
 
fig.suptitle(f'1st-order RC HPF (R = 1 k$\\Omega$, C = 0.01 $\\mu$F, '
             f'$f_c$ = {fc:.0f} Hz, $V_i$ = {V_in_V_PP:g} V$_{{pp}}$)')
 
# (0) Output amplitude, linear-linear
ax0.plot(f_th_lin, V_in_V_PP * H_th_lin, '-', label='Theory')
ax0.plot(f_Hz, V_V_PP, 'o', label='Measured')
ax0.axvline(fc, color='gray', ls='--', lw=1)
ax0.axhline(V_in_V_PP / np.sqrt(2), color='gray', ls=':', lw=1)
ax0.set_xlim(0, f_Hz.max() * 1.05)
ax0.set_ylim(0, None)
ax0.set_xlabel('Frequency [Hz]')
ax0.set_ylabel('$V_o$ [V$_{pp}$]')
ax0.set_title('Linear scale')
ax0.grid(True, ls=':', alpha=0.6)
ax0.legend(loc='lower right')
 
# (1) Output amplitude, log-log
ax1.loglog(f_th, V_in_V_PP * H_th, '-', label='Theory')
ax1.loglog(f_Hz, V_V_PP, 'o', label='Measured')
ax1.axvline(fc, color='gray', ls='--', lw=1)
ax1.axhline(V_in_V_PP / np.sqrt(2), color='gray', ls=':', lw=1)
ax1.set_ylabel('$V_o$ [V$_{pp}$]')
ax1.set_title('Log-log scale')
ax1.grid(True, which='both', ls=':', alpha=0.6)
ax1.legend(loc='lower right')
ax1.tick_params(labelbottom=False)
 
# (2) Gain in dB, log frequency (Bode magnitude)
# Asymptotic (straight-line) Bode approximation:
#   f < fc : +20 dB/dec line through 0 dB at fc  -> 20*log10(f/fc)
#   f >= fc: 0 dB
gain_asym_dB = np.minimum(20 * np.log10(f_th / fc), 0)
 
ax2.semilogx(f_th, gain_th_dB, '-', label='Theory')
ax2.semilogx(f_th, gain_asym_dB, '-', color='red', lw=1.5,
             label='Straight-line approx.')
ax2.semilogx(f_Hz, gain_meas_dB, 'o', label='Measured')
ax2.axvline(fc, color='gray', ls='--', lw=1, label=f'$f_c$ = {fc:.0f} Hz')
ax2.axhline(-3, color='gray', ls=':', lw=1, label='-3 dB')
ax2.set_xlabel('Frequency [Hz]')
ax2.set_ylabel('Gain [dB]')
ax2.set_title('Bode magnitude (semilog-x)')
ax2.grid(True, which='both', ls=':', alpha=0.6)
ax2.legend(loc='lower right')
 
plt.tight_layout()
# plt.savefig('hpf_freq_response.png', dpi=150)
plt.show()
```

```python
# f_Hz =   [    ??,    ??,    ??,    ??,    ??,    ??,    ??,    ??,    ??]
# V_V_PP = [    ??,    ??,    ??,    ??,    ??,    ??,    ??,    ??,    ??]
```

```python
f_Hz = np.array(f_Hz)
V_V_PP = np.array(V_V_PP)

# ---- Circuit parameters (inverting active HPF) ----
R1 = 1 * 1000         # ohm
C1 = 0.015 / 1000000  # F
R2 = 10 * 1000        # ohm
K = R2 / R1                          # passband gain = 10 (20 dB)
fc = 1 / (2 * np.pi * R1 * C1)       # ~10.6 kHz
K_dB = 20 * np.log10(K)

# Input amplitude (Vpp)
V_in_V_PP = 1.0

# Op-amp slew rate [V/s], used only for the slew-rate limit curve.
# 0.5 V/us is the typical value for a uA741.
SHOW_SLEW_LIMIT = True
SR = 0.5e6
 
# ---- Theoretical curve ----
# |H| = (R2/R1) * (f/fc) / sqrt(1 + (f/fc)^2)
def H_hpf(f):
    x = f / fc
    return K * x / np.sqrt(1 + x ** 2)
 
# Maximum Vpp a slew-limited op amp can output at frequency f
# (full-slew triangle wave: Vpp = SR * T/2 = SR / (2f))
def V_slew_pp(f):
    return SR / (2 * f)
 
f_th = np.logspace(np.log10(f_Hz.min() / 2), np.log10(f_Hz.max() * 2), 500)
H_th = H_hpf(f_th)
 
gain_meas_dB = 20 * np.log10(V_V_PP / V_in_V_PP)
gain_th_dB = 20 * np.log10(H_th)
 
# Straight-line (asymptotic) Bode approximation:
#   f <  fc: +20 dB/dec line reaching 20*log10(K) at fc
#   f >= fc: 20*log10(K)
gain_asym_dB = K_dB + np.minimum(20 * np.log10(f_th / fc), 0)
 
# Theoretical curve for the linear plot (evenly spaced on a linear axis)
f_th_lin = np.linspace(0, f_Hz.max() * 1.05, 1000)
H_th_lin = H_hpf(f_th_lin)
 
# Slew-rate limit curve (drawn only above 5 kHz so it stays in the plot area)
f_sr = f_th[f_th > 5e3]
f_sr_lin = f_th_lin[f_th_lin > 5e3]
 
# ---- Plot ----
# ax0: linear axes (x is not shared, because a shared x axis must use one scale)
# ax1, ax2: log frequency axis, shared
fig = plt.figure(figsize=(8, 12))
ax0 = fig.add_subplot(3, 1, 1)
ax1 = fig.add_subplot(3, 1, 2)
ax2 = fig.add_subplot(3, 1, 3, sharex=ax1)
 
fig.suptitle(f'Inverting active HPF (R$_1$ = 1 k$\\Omega$, C$_1$ = 0.015 $\\mu$F, '
             f'R$_2$ = 10 k$\\Omega$, $f_c$ = {fc:.0f} Hz, '
             f'$V_i$ = {V_in_V_PP:g} V$_{{pp}}$)', fontsize=11)
 
sr_label = f'Slew-rate limit (SR = {SR / 1e6:g} V/$\\mu$s)'
 
# (0) Output amplitude, linear-linear
ax0.plot(f_th_lin, V_in_V_PP * H_th_lin, '-', label='Theory')
if SHOW_SLEW_LIMIT:
    ax0.plot(f_sr_lin, V_slew_pp(f_sr_lin), '-.', color='purple', lw=1.2,
             label=sr_label)
ax0.plot(f_Hz, V_V_PP, 'o', color='C1', label='Measured')
ax0.axvline(fc, color='gray', ls='--', lw=1)
ax0.axvline(f_0, color='C1', ls='--', lw=1)
ax0.axhline(V_out_max_V_PP / np.sqrt(2), color='gray', ls=':', lw=1)
ax0.set_xlim(0, f_Hz.max() * 1.05)
ax0.set_ylim(0, K * V_in_V_PP * 1.15)
ax0.set_xlabel('Frequency [Hz]')
ax0.set_ylabel('$V_o$ [V$_{pp}$]')
ax0.set_title('Linear scale')
ax0.grid(True, ls=':', alpha=0.6)
ax0.legend(loc='lower center')
 
# (1) Output amplitude, log-log
ax1.loglog(f_th, V_in_V_PP * H_th, '-', label='Theory')
if SHOW_SLEW_LIMIT:
    ax1.loglog(f_sr, V_slew_pp(f_sr), '-.', color='purple', lw=1.2,
               label=sr_label)
ax1.loglog(f_Hz, V_V_PP, 'o', color='C1', label='Measured')
ax1.axvline(fc, color='gray', ls='--', lw=1)
ax1.axvline(f_0, color='C1', ls='--', lw=1)
ax1.axhline(V_out_max_V_PP / np.sqrt(2), color='gray', ls=':', lw=1)
ax1.set_ylim(V_V_PP.min() / 3, K * V_in_V_PP * 3)
ax1.set_ylabel('$V_o$ [V$_{pp}$]')
ax1.set_title('Log-log scale')
ax1.grid(True, which='both', ls=':', alpha=0.6)
ax1.legend(loc='lower right')
ax1.tick_params(labelbottom=False)
 
# (2) Gain in dB, log frequency (Bode magnitude)
ax2.semilogx(f_th, gain_th_dB, '-', label='Theory')
ax2.semilogx(f_th, gain_asym_dB, '-', color='red', lw=1.5,
             label='Straight-line approx.')
if SHOW_SLEW_LIMIT:
    ax2.semilogx(f_sr, 20 * np.log10(V_slew_pp(f_sr) / V_in_V_PP), '-.',
                 color='purple', lw=1.2, label=sr_label)
ax2.semilogx(f_Hz, gain_meas_dB, 'o', color='C1', label='Measured')
ax2.axvline(fc, color='gray', ls='--', lw=1, label=f'$f_c$ (theory) = {fc:.0f} Hz')
ax2.axvline(f_0, color='C1', ls='--', lw=1, label=f'$f_0$ (measured) = {f_0:g} Hz')
ax2.axhline(K_dB - 3, color='gray', ls=':', lw=1, label=f'{K_dB - 3:.0f} dB (max - 3 dB)')
ax2.set_ylim(gain_meas_dB.min() - 8, K_dB + 8)
ax2.set_xlabel('Frequency [Hz]')
ax2.set_ylabel('Gain [dB]')
ax2.set_title('Bode magnitude (semilog-x)')
ax2.grid(True, which='both', ls=':', alpha=0.6)
ax2.legend(loc='lower right', fontsize=9)
 
plt.tight_layout()
# plt.savefig('active_hpf_freq_response.png', dpi=150)
plt.show()
```

```python
f_Hz = np.array(f_Hz)
V_V_PP = np.array(V_V_PP)
 
# ---- Circuit parameters (equal-component Sallen-Key 2nd-order LPF) ----
R = 10e3         # ohm  (R1 = R2 = R)
C = 0.015e-6     # F    (C1 = C2 = C)
RA = 1e3         # ohm
RB = 600         # ohm
 
K = 1 + RB / RA                    # passband gain = 1.6
f0 = 1 / (2 * np.pi * R * C)       # natural frequency ~1061 Hz
Q = 1 / (3 - K)                    # quality factor ~0.714
K_dB = 20 * np.log10(K)

# Input amplitude (Vpp)
V_in_V_PP = 1.0

# ---- Theoretical curve ----
# H = K / (1 - (f/f0)^2 + j (f/f0)/Q)
# |H| = K / sqrt((1 - x^2)^2 + (x/Q)^2),  x = f/f0
def H_lpf2(f):
    x = f / f0
    return K / np.sqrt((1 - x ** 2) ** 2 + (x / Q) ** 2)
 
f_th = np.logspace(np.log10(f_Hz.min() / 2), np.log10(f_Hz.max() * 2), 500)
H_th = H_lpf2(f_th)
 
gain_meas_dB = 20 * np.log10(V_V_PP / V_in_V_PP)
gain_th_dB = 20 * np.log10(H_th)
 
# Straight-line (asymptotic) Bode approximation:
#   f <= f0: 20*log10(K)
#   f >  f0: -40 dB/dec line from 20*log10(K) at f0
gain_asym_dB = K_dB + np.minimum(-40 * np.log10(f_th / f0), 0)
 
# Theoretical curve for the linear plot (evenly spaced on a linear axis)
f_th_lin = np.linspace(0, f_Hz.max() * 1.05, 1000)
H_th_lin = H_lpf2(f_th_lin)
 
# ---- Plot ----
# ax0: linear axes (x is not shared, because a shared x axis must use one scale)
# ax1, ax2: log frequency axis, shared
fig = plt.figure(figsize=(8, 12))
ax0 = fig.add_subplot(3, 1, 1)
ax1 = fig.add_subplot(3, 1, 2)
ax2 = fig.add_subplot(3, 1, 3, sharex=ax1)
 
fig.suptitle(f'Sallen-Key 2nd-order LPF (R = 10 k$\\Omega$, C = 0.015 $\\mu$F, '
             f'R$_A$ = 1 k$\\Omega$, R$_B$ = 600 $\\Omega$)\n'
             f'K = {K:g}, $f_0$ = {f0:.0f} Hz, Q = {Q:.3f}, '
             f'$V_i$ = {V_in_V_PP:g} V$_{{pp}}$', fontsize=11)
 
# (0) Output amplitude, linear-linear
ax0.plot(f_th_lin, V_in_V_PP * H_th_lin, '-', label='Theory')
ax0.plot(f_Hz, V_V_PP, 'o', label='Measured')
ax0.axvline(f0, color='gray', ls='--', lw=1)
ax0.axhline(K * V_in_V_PP / np.sqrt(2), color='gray', ls=':', lw=1)
ax0.set_xlim(0, f_Hz.max() * 1.05)
ax0.set_ylim(0, None)
ax0.set_xlabel('Frequency [Hz]')
ax0.set_ylabel('$V_o$ [V$_{pp}$]')
ax0.set_title('Linear scale')
ax0.grid(True, ls=':', alpha=0.6)
ax0.legend()
 
# (1) Output amplitude, log-log
ax1.loglog(f_th, V_in_V_PP * H_th, '-', label='Theory')
ax1.loglog(f_Hz, V_V_PP, 'o', label='Measured')
ax1.axvline(f0, color='gray', ls='--', lw=1)
ax1.axhline(K * V_in_V_PP / np.sqrt(2), color='gray', ls=':', lw=1)
ax1.set_ylabel('$V_o$ [V$_{pp}$]')
ax1.set_title('Log-log scale')
ax1.grid(True, which='both', ls=':', alpha=0.6)
ax1.legend(loc='lower left')
ax1.tick_params(labelbottom=False)
 
# (2) Gain in dB, log frequency (Bode magnitude)
ax2.semilogx(f_th, gain_th_dB, '-', label='Theory')
ax2.semilogx(f_th, gain_asym_dB, '-', color='red', lw=1.5,
             label='Straight-line approx. (-40 dB/dec)')
ax2.semilogx(f_Hz, gain_meas_dB, 'o', label='Measured')
ax2.axvline(f0, color='gray', ls='--', lw=1, label=f'$f_0$ = {f0:.0f} Hz')
ax2.axhline(K_dB - 3, color='gray', ls=':', lw=1,
            label=f'{K_dB - 3:.2f} dB (max - 3 dB)')
ax2.set_ylim(gain_meas_dB.min() - 10, K_dB + 6)
ax2.set_xlabel('Frequency [Hz]')
ax2.set_ylabel('Gain [dB]')
ax2.set_title('Bode magnitude (semilog-x)')
ax2.grid(True, which='both', ls=':', alpha=0.6)
ax2.legend(loc='lower left')
 
plt.tight_layout()
# plt.savefig('sallen_key_lpf_freq_response.png', dpi=150)
plt.show()
```
