### 실험07 L298보드의 이해

L298 보드

```python
display_image("./images/ch13_05_ep07_01_parts_01.png", width=800)
```

* L298보드는 2가지 전원이 필요하다.
    + 1) 모터구동용 : 5-50V
    + 2)L298보드의 논리회로 전원용 : +5V
* 논리회로 전원용을 위한 +5V를 공급하는데는 2가지 방법이 있다.
    + 1) 외부에서 +5V를 공급하는 방법
    + 2) 내부에서 +5V를 만드는 방법
* 외부에서 +5V를 공급하는 경우에는
    + 6번 위치의 "5V Enable" 점퍼스위치를 제거하면 된다.
* 내부에서 +5V를 만드는 경우는 내부에 내장된 레귤레이터가 모터구동용 전원을 이용하여 +5V를 만든다.
    + 6번 위치의 "5V Enable" 점퍼스위치를 연결하면 된다.
    + 레귤레이터의 입력전압(모터구동용 전원)이 대략 5~12V정도일 때 사용하는 것이 적당하다.
    + **우리는 모터구동전압으로 약 8V를 사용할 것이므로, 이 점퍼를 연결한 상태로 이용한다.**
