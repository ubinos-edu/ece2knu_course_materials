* 부하      : 무한대 (High-Z)
* 종류      : 펄스파
* 주기      : 1 초
* 최대      : 3.3 V
* 최소      : 0 V
* 듀티사이클: 50 %

* 펄스파 설정 방법
  + Waveform 버튼
        - Pulse
    + Units 버튼
        - Period
        - High/Low
        - Duty Cycle
    + Parameters 버튼
        - Period: 1초
        - High level: 3.3 V
        - Low level: 0 V
        - Duty cycle: 50 %
        - Lead Edge: 10 ns
        - Trail Edge: 10 ns


[결과] 1초를 주기를 LED가 켜짐과 꺼짐을 반복한다

주기를 다음과 같이 줄여보자.

추가 단락 검사

회로도 (우)의 (1)과 (2)번을 단락 검사를 한다.

[결과] 단락이다.

[퀴즈] 왜 결과가 단락으로 나오는가?

* 릴레이 내부의 코일로 연결되어 있고, 이 코일의 저항이 대략 50 Ohm 정도이다. DMM(Fluke) 은 이 정도의 저항을 단락으로 판단하는 것으로 보인다.

DMM(GoldStar)로 단락 검사를 한다.

[결과] 소리가 나지 않는다. (단락이 아닌 것으로 판단)

[퀴즈] 이 경우에는 왜 단락으로 나오지 않는가?

* DMM(GoldStar) 은 50 Ohm 정도의 저항을 단악으로 판단하지 않는 것으로 보인다. (단락 여부 판단 기준은 DMM 제조사마다 다르다.)

DMM의 다이오드 검사모드로 LED를 검사한다.

* 회로도 (우)의 (6)에 빨간색 리드선을 연결하고, (5)에 검은색 리드선을 연결한다.

[결과] LED가 ON이 된다.

릴레이보드에 전원을 공급한다.

* 전원공급: V+=5V, V-=접지
* 점퍼선으로 연결한다. 한쪽은 수커넥터이고, 다른 쪽은 암커넥터인 점퍼선을 이용한다.

입력 v_i 에 신호를 인가하자 (파형발생기를 이용)

* 부하  : 무한대 (High-Z)
* 종류  : 구형파
