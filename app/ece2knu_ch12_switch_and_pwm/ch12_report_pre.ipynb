* 사용 소자
    + NPN TR: Place Symbols > Simulation_SPICE > NPN
        - Simulation Model: 2N2222A_beta_250.MOD > Q2n222a
    + LED Diode: Place Symbols > Device > LED
        - Simulation Model: led_models.lib > LED_Red
        - **Pin assignment 도 맞춰줘야 함**
    + VIN: Place Symbols > Simulation_SPICE > VPULSE
        - Sim Params: y1=3.3 y2=0 td=2n tr=10n tf=10n tw=500m per=1000m
